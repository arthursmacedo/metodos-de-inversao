```python
import numpy as np
import matplotlib.pyplot as plt
```

Ajustar função de reta: estimar coeficiente angular "a" e coeficiente linear "b" a partir de um conjunto de dados sintéticos (yo,x)

```python
a = 42 #coeficiente angular verdadeiro
b = 10_000 #coeficiente linear verdadeiro
N = 100 #numero de pontos

# Criar conjunto de dados yo,x com erro em yo (y observado)
random = np.random.default_rng()
x = random.uniform(0, 1_000, N)
yo = a * x + b
errors = random.normal(0, 0.05 * np.abs(yo).max(), N)
yo = yo + errors

```

```python
# Plota os dados criados

plt.plot(x, yo, ".")
```

```python
# Estimar coeficientes "a" e "b" pela solução do MMQ
a_est = (N * np.sum(yo * x) - np.sum(x) * np.sum(yo)) / (N * np.sum(x**2) - np.sum(x)**2)
b_est = (np.sum(yo) - a_est * np.sum(x)) / N
print(a_est, b_est)
```

```python
# Agora plotamos nossa solução por MMQ
plt.plot(x, yo, ".")
y_est = a_est * x + b_est
plt.plot(x, y_est,)
```
